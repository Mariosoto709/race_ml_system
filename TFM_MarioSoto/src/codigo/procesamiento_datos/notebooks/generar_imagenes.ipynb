```python
import os
import matplotlib.pyplot as plt
from PIL import Image

def plot_selected_pngs(folder_path, image_names, figsize=(18, 6)):
    """
    Plotea horizontalmente PNGs seleccionados de una carpeta.

    Parameters
    ----------
    folder_path : str
        Ruta a la carpeta.
    image_names : list of str
        Nombres de los archivos PNG a mostrar.
    figsize : tuple
        Tamaño de la figura.
    """
    image_paths = [os.path.join(folder_path, name) for name in image_names]

    n = len(image_paths)
    fig, axes = plt.subplots(1, n, figsize=figsize)

    if n == 1:
        axes = [axes]

    for ax, img_path in zip(axes, image_paths):
        img = Image.open(img_path)
        ax.imshow(img)
        ax.axis("off")

    plt.tight_layout()
    plt.show()


# Carpeta
folder = r"C:\Users\mario\Downloads"

# PNGs que TÚ eliges
selected_images = [
    "xgb_curva_tiempo_mar.png",
    "lgbm_curva_tiempo_mar.png",
]

plot_selected_pngs(folder, selected_images)
```
